```python
price_cols = [
    "promo_offer_price",
    "product_price",
    "item_selling_price",
    "item_mrp"
]

for col in price_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("₹", "", regex=False)
        .str.replace("$", "", regex=False)
        .str.replace("€", "", regex=False)
        .str.replace(",", "", regex=False)
    )

    df[col] = pd.to_numeric(df[col], errors="coerce")
```

### Stores with poor inventory availability

```python
type("out_of_stock_flag")
```

```python
df["out_of_stock_numeric"] = df["out_of_stock_flag"].map({
    "Yes": 1,
    "No": 0
})
```

```python
out_stock = (df.groupby("store_name_location")["out_of_stock_numeric"]
             .mean()
             .mul(100)
             .round(2)
             .astype(str) + "%").reset_index()

out_stock
```

```python
import textwrap

out_stock["store_short"] = out_stock["store_name_location"].apply(
    lambda x: "\n".join(textwrap.wrap(x, width=25))
)
```

```python
out_stock["store_short"] = (
    out_stock["store_name_location"]
    .str.split(",")
    .str[0]
)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=out_stock,
    y="store_short",
    x="out_of_stock_numeric"
)

plt.xlabel("Out-of-Stock Rate (%)")
plt.ylabel("Store")
plt.title("Out-of-Stock Percentage by Store")

plt.tight_layout()
plt.show()
```

### Key Insights

- **Trends,MP** has the highest out-of-stock rate (**61.0%**), followed by **Shree Umiya Motors,Maharashtra** (**57.9%**) and **ShivSai Complex,Maharashtra** (**50.0%**).
- Among major stores, **Croma,UP** records a relatively high stockout rate (**30.7%**), while **Blinkit Store HP-2** maintains the lowest (**13.7%**).
- **Purohit Sweet Mart,Maharashtra**, **Sapna Muskan Photography,Delhi**, and **King Uncle,MP** show moderate stockout levels ranging from **16% to 20%**.
- Several small stores, including **Anu Tea Stall,UP**, **Best Homeopathic Doctor,Maharashtra**, and **Carver Public School,UP**, report **0% out-of-stock products**.
